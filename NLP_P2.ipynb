```python
#demo1

import re
from collections import Counter

teks = "Saya suka belajar NLP. NLP membuat komputer memahami bahasa manusia, dan saya suka bahasa Indonesia."
token = re.findall(r"[a-zA-Z]+", teks.lower())
print(token)
print(Counter(token).most_common(5))
```

```python
#demo2

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

faq = ["Bagaimana cara mengisi KRS?",
       "Kapan jadwal pembayaran UKT?",
       "Bagaimana mendaftar beasiswa?"]
tanya = "cara isi KRS online"

vec = TfidfVectorizer()
X = vec.fit_transform(faq + [tanya])
skor = cosine_similarity(X[-1], X[:-1])[0]
print([round(float(s), 2) for s in skor])
print("Jawaban terdekat:", faq[skor.argmax()])
```

```python
#demo4

faq_s = [stem(f.lower()) for f in faq]
q = stem(tanya.lower())
print(faq_s, "|", q)

vec = TfidfVectorizer()
X = vec.fit_transform(faq_s + [q])
skor = cosine_similarity(X[-1], X[:-1])[0]
print([round(float(s), 2) for s in skor])
print("Jawaban terdekat:", faq[skor.argmax()])
```

```python
#demo5

from sklearn.metrics import accuracy_score, recall_score, f1_score

y_benar = [1]*5 + [0]*95      # 5 spam, 95 bukan spam
y_tebak = [0]*100             # model selalu menebak "bukan spam"
print("Akurasi:", accuracy_score(y_benar, y_tebak))
print("Recall spam:", recall_score(y_benar, y_tebak))
print("F1 spam:", f1_score(y_benar, y_tebak, zero_division=0))
```
